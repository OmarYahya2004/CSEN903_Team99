```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import os
from IPython.display import display

# Set visualization style
sns.set_theme(style="whitegrid")

# Automatically locate all CSV files in the Kaggle input directory
input_dir = '/kaggle/input/'
dataset_paths = []
for root, dirs, files in os.walk(input_dir):
    for file in files:
        if file.endswith('.csv'):
            dataset_paths.append(os.path.join(root, file))

# Dictionary to store all dataframes for later multi-hop joins
f1_data = {}
raw_f1_data = {} # ADDED: Dictionary for raw data

print(f"Found {len(dataset_paths)} F1 CSV files. Beginning automated EDA...")

# Loop through and visualize each table
for path in dataset_paths:
    file_name = os.path.basename(path)
    table_name = file_name.replace('.csv', '')
    
    # ADDED: Load RAW data without parsing '\N' (Immutable Raw Copy for Before-Cleaning EDA)
    raw_df = pd.read_csv(path, dtype=str, keep_default_na=False)
    raw_f1_data[table_name] = raw_df
    
    # Load data, converting the '\N' sentinel string into true NaN values
    df = pd.read_csv(path, na_values=['\\N'])
    f1_data[table_name] = df
    
    # 1. Structural Audit
    print(f"\n\n{'='*80}")
    print(f"TABLE: {file_name.upper()} | Shape: {df.shape[0]:,} rows, {df.shape[1]} columns")
    print(f"{'='*80}")
    display(df.head(3))
    
    # 2. Missing Values Visualization (Audit Requirement)
    missing_count = df.isna().sum().sum()
    if missing_count > 0:
        plt.figure(figsize=(12, 2))
        sns.heatmap(df.isna(), cbar=False, cmap='viridis', yticklabels=False)
        plt.title(f"Missing Values Heatmap: {table_name} ({missing_count:,} total missing)", pad=10)
        plt.show()
    else:
        print(f"✓ No missing values detected in {table_name}.")
        
    # 3. Numerical Distributions (EDA Requirement)
    num_cols = df.select_dtypes(include=['float64', 'int64']).columns
    if len(num_cols) > 0:
        # Calculate grid size based on number of numerical columns
        cols_count = len(num_cols)
        fig_width = min(15, cols_count * 3)
        
        df[num_cols].hist(figsize=(fig_width, 3), bins=30, layout=(1, cols_count), edgecolor='black')
        plt.suptitle(f"Numerical Distributions: {table_name}", y=1.1, fontweight='bold')
        plt.tight_layout()
        plt.show()
    else:
        print(f"ℹ No continuous numerical columns to plot for {table_name}.")
```


# Cell 2: Building The Main Table And Visualizing The Data After Cleaing

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print("=" * 100)
print("             OBJECTIVE 1: DATA CLEANING, PK/FK & GRAIN AUDIT")
print("=" * 100)

################################
# 1. Base Table & Grain Audit (results.csv)
################################
results = f1_data['results'].copy()

# --- 1A. Primary Key Audit (All 8 Training Tables) ---
print("\n[1] PRIMARY KEY AUDIT (Uniqueness & Non-Null Check)")
print("-" * 100)
pk_tables = [
    ('results', 'resultId'), ('races', 'raceId'), ('drivers', 'driverId'), 
    ('constructors', 'constructorId'), ('circuits', 'circuitId'), ('qualifying', 'qualifyId'),
    ('driver_standings', 'driverStandingsId'), ('constructor_standings', 'constructorStandingsId')
]
for tbl_name, pk_col in pk_tables:
    tbl = f1_data[tbl_name]
    dups = tbl[pk_col].duplicated().sum()
    nulls = tbl[pk_col].isna().sum()
    print(f"  • {tbl_name:<22} | PK: {pk_col:<22} | Dups: {dups} | Nulls: {nulls}")

# --- 1B. Foreign Key Audit (All Relationships Across Training Tables) ---
print("\n[2] FOREIGN KEY RECONCILIATION AUDIT (Orphan Check Across All Training Tables)")
print("-" * 100)
fk_checks = [
    ("results.raceId -> races.raceId", (~f1_data['results']['raceId'].isin(f1_data['races']['raceId'])).sum()),
    ("results.driverId -> drivers.driverId", (~f1_data['results']['driverId'].isin(f1_data['drivers']['driverId'])).sum()),
    ("results.constructorId -> constructors.constructorId", (~f1_data['results']['constructorId'].isin(f1_data['constructors']['constructorId'])).sum()),
    ("races.circuitId -> circuits.circuitId", (~f1_data['races']['circuitId'].isin(f1_data['circuits']['circuitId'])).sum()),
    ("qualifying.raceId -> races.raceId", (~f1_data['qualifying']['raceId'].isin(f1_data['races']['raceId'])).sum()),
    ("qualifying.driverId -> drivers.driverId", (~f1_data['qualifying']['driverId'].isin(f1_data['drivers']['driverId'])).sum()),
    ("qualifying.constructorId -> constructors.constructorId", (~f1_data['qualifying']['constructorId'].isin(f1_data['constructors']['constructorId'])).sum()),
    ("driver_standings.raceId -> races.raceId", (~f1_data['driver_standings']['raceId'].isin(f1_data['races']['raceId'])).sum()),
    ("driver_standings.driverId -> drivers.driverId", (~f1_data['driver_standings']['driverId'].isin(f1_data['drivers']['driverId'])).sum()),
    ("constructor_standings.raceId -> races.raceId", (~f1_data['constructor_standings']['raceId'].isin(f1_data['races']['raceId'])).sum()),
    ("constructor_standings.constructorId -> constructors.constructorId", (~f1_data['constructor_standings']['constructorId'].isin(f1_data['constructors']['constructorId'])).sum()),
]
for rel_desc, orphans in fk_checks:
    print(f"  • {rel_desc:<65} | Orphans: {orphans}")

# --- 1C. Target Creation & Shared-Drive Grain Deduplication ---
results['podium'] = (results['positionOrder'] <= 3).astype(int)

# Track exact metrics BEFORE deduplication for reporting & plotting
raw_results_rows = len(results)
dup_mask = results.duplicated(subset=['raceId', 'driverId'], keep=False)
dup_rows_before = dup_mask.sum()
extra_dup_rows_dropped = results.duplicated(subset=['raceId', 'driverId'], keep='first').sum()

# ADDED MISSING AUDIT LOGIC HERE: Quantify unique pairs and differing outcomes
unique_duplicate_pairs = results[dup_mask][['raceId', 'driverId']].drop_duplicates().shape[0]
conflict_df = results[dup_mask].groupby(['raceId', 'driverId'])['podium'].nunique().reset_index()
differing_outcomes = (conflict_df['podium'] > 1).sum()

# Preserve podium / best finish if driver shared multiple cars, then keep starting car (best non-zero grid)
results['podium'] = results.groupby(['raceId', 'driverId'])['podium'].transform('max')
results['positionOrder'] = results.groupby(['raceId', 'driverId'])['positionOrder'].transform('min')

results.sort_values(
    by=['raceId', 'driverId', 'grid'],
    key=lambda col: col.replace(0, 999) if col.name == 'grid' else col,
    inplace=True
)
results.drop_duplicates(subset=['raceId', 'driverId'], keep='first', inplace=True)

clean_results_rows = len(results)
dup_rows_after = results.duplicated(subset=['raceId', 'driverId'], keep=False).sum()

# Note: Keep positionOrder temporarily for Cell 5 rolling form, drop other post-race leaky columns
leakage_cols = ['points', 'positionText', 'position', 'laps', 'time', 'milliseconds', 
                'fastestLap', 'rank', 'fastestLapTime', 'fastestLapSpeed', 'statusId', 'number']
results.drop(columns=leakage_cols, inplace=True, errors='ignore')

################################
# 2. Chronological Spine (races.csv)
################################
races = f1_data['races'].copy()
races['date'] = pd.to_datetime(races['date'], errors='coerce')
race_drop_cols = ['url', 'fp1_date', 'fp1_time', 'fp2_date', 'fp2_time', 'fp3_date', 
                  'fp3_time', 'quali_date', 'quali_time', 'sprint_date', 'sprint_time', 'time']
races.drop(columns=race_drop_cols, inplace=True, errors='ignore')
races.rename(columns={'name': 'race_name'}, inplace=True)

################################
# 3. Qualifying Data (qualifying.csv) - Leave NaNs for SimpleImputer!
################################
qualifying = f1_data['qualifying'].copy()
def time_to_seconds(t):
    if pd.isna(t) or t == '\\N': return np.nan
    try:
        parts = str(t).split(':')
        return float(parts[0]) * 60 + float(parts[1]) if len(parts) == 2 else float(parts[0])
    except:
        return np.nan

for col in ['q1', 'q2', 'q3']:
    qualifying[col] = qualifying[col].apply(time_to_seconds)

qualifying.rename(columns={'position': 'quali_pos'}, inplace=True)
qualifying.sort_values(by=['raceId', 'driverId', 'quali_pos'], inplace=True)
qualifying_dups_before = qualifying.duplicated(subset=['raceId', 'driverId'], keep=False).sum()
qualifying.drop_duplicates(subset=['raceId', 'driverId'], keep='first', inplace=True)
qualifying_dups_after = qualifying.duplicated(subset=['raceId', 'driverId'], keep=False).sum()
qualifying.drop(columns=['qualifyId', 'number', 'constructorId'], inplace=True, errors='ignore')

################################
# 4. Driver, Constructor & Circuit Data
################################
drivers = f1_data['drivers'].copy()
drivers['dob'] = pd.to_datetime(drivers['dob'], errors='coerce')
drivers.drop(columns=['url', 'number', 'code'], inplace=True, errors='ignore')

constructors = f1_data['constructors'].copy()
constructors.drop(columns=['url'], inplace=True, errors='ignore')
constructors.rename(columns={'name': 'constructor_name', 'nationality': 'constructor_nationality'}, inplace=True)

circuits = f1_data['circuits'].copy()
circuits['alt'] = pd.to_numeric(circuits['alt'], errors='coerce')
circuits.drop(columns=['url'], inplace=True, errors='ignore')
circuits.rename(columns={'name': 'circuit_name', 'location': 'circuit_location', 'country': 'circuit_country'}, inplace=True)

################################
# 5. Standings (Grouped by ['driverId', 'year'] so Round 1 starts at 0 pts!)
################################
driver_stnd = f1_data['driver_standings'].merge(races[['raceId', 'year', 'round', 'date']], on='raceId')
driver_stnd.sort_values(by=['driverId', 'year', 'round'], inplace=True)
driver_stnd['driver_pts_pre_race'] = driver_stnd.groupby(['driverId', 'year'])['points'].shift(1).fillna(0)
driver_stnd['driver_pos_pre_race'] = driver_stnd.groupby(['driverId', 'year'])['position'].shift(1).fillna(99)
driver_stnd_dups_before = driver_stnd.duplicated(subset=['raceId', 'driverId'], keep=False).sum()
driver_stnd = driver_stnd.drop_duplicates(subset=['raceId', 'driverId'])[['raceId', 'driverId', 'driver_pts_pre_race', 'driver_pos_pre_race']]
driver_stnd_dups_after = driver_stnd.duplicated(subset=['raceId', 'driverId'], keep=False).sum()

const_stnd = f1_data['constructor_standings'].merge(races[['raceId', 'year', 'round', 'date']], on='raceId')
const_stnd.sort_values(by=['constructorId', 'year', 'round'], inplace=True)
const_stnd['constructor_pts_pre_race'] = const_stnd.groupby(['constructorId', 'year'])['points'].shift(1).fillna(0)
const_stnd['constructor_pos_pre_race'] = const_stnd.groupby(['constructorId', 'year'])['position'].shift(1).fillna(99)
const_stnd_dups_before = const_stnd.duplicated(subset=['raceId', 'constructorId'], keep=False).sum()
const_stnd = const_stnd.drop_duplicates(subset=['raceId', 'constructorId'])[['raceId', 'constructorId', 'constructor_pts_pre_race', 'constructor_pos_pre_race']]
const_stnd_dups_after = const_stnd.duplicated(subset=['raceId', 'constructorId'], keep=False).sum()

################################
# 6. The Grand Join
################################
df = results.merge(races, on='raceId', how='left')
df = df.merge(circuits, on='circuitId', how='left')
df = df.merge(drivers, on='driverId', how='left')
df['driver_age'] = (df['date'] - df['dob']).dt.days / 365.25
df.drop(columns=['dob'], inplace=True) # Keep 'date' for Cell 5 chronological sorting!

df = df.merge(constructors, on='constructorId', how='left')
df = df.merge(qualifying, on=['raceId', 'driverId'], how='left')
df = df.merge(driver_stnd, on=['raceId', 'driverId'], how='left')
df = df.merge(const_stnd, on=['raceId', 'constructorId'], how='left')
df.drop(columns=['resultId'], inplace=True, errors='ignore')

################################
# 7. Post-Merge Cleanup (ML-Safe)
################################
df['driver_pts_pre_race'] = df['driver_pts_pre_race'].fillna(0)
df['constructor_pts_pre_race'] = df['constructor_pts_pre_race'].fillna(0)
df['driver_pos_pre_race'] = df['driver_pos_pre_race'].fillna(99)
df['constructor_pos_pre_race'] = df['constructor_pos_pre_race'].fillna(99)
df['quali_pos'] = df['quali_pos'].fillna(99)

assert df.duplicated(subset=['raceId', 'driverId']).sum() == 0, "Grain violation!"

print("\n[3] GRAIN AUDIT: (raceId, driverId) UNIQUENESS & SHARED DRIVES")
print("-" * 100)
print(f"  • results.csv rows BEFORE deduplication     : {raw_results_rows:,}")
print(f"  • Rows involved in shared-drive duplicates  : {dup_rows_before:,} rows ({extra_dup_rows_dropped:,} redundant rows dropped)")
print(f"  • Unique (raceId, driverId) conflicting pairs: {unique_duplicate_pairs}")
print(f"  • Pairs with conflicting Podium targets     : {differing_outcomes}")
print(f"  • results.csv rows AFTER deduplication      : {clean_results_rows:,}")
print(f"  • Duplicate (raceId, driverId) in results   : {dup_rows_before} -> {dup_rows_after}")
print(f"  • Duplicate (raceId, driverId) in qualifying: {qualifying_dups_before} -> {qualifying_dups_after}")
print(f"  • Duplicate (raceId, driverId) in final df  : {df.duplicated(subset=['raceId', 'driverId']).sum()}")
print(f"  • Rule Applied                              : Kept starting car (lowest non-zero grid), preserved max(podium)")

print("\n[4] TABLE SHAPE & MISSINGNESS SUMMARY (BEFORE vs. AFTER CLEANING)")
print("-" * 100)
cleaned_tables = {
    'results': results,
    'races': races,
    'qualifying': qualifying,
    'drivers': drivers,
    'constructors': constructors,
    'circuits': circuits,
    'driver_standings': driver_stnd,
    'constructor_standings': const_stnd
}
print(f"  {'Table Name':<23} | {'Raw Shape':<16} | {'Cleaned Shape':<16} | {'Remaining Missing'}")
print("-" * 100)
for name, clean_tbl in cleaned_tables.items():
    raw_tbl = f1_data[name]
    raw_shape = f"{raw_tbl.shape[0]:,} x {raw_tbl.shape[1]}"
    clean_shape = f"{clean_tbl.shape[0]:,} x {clean_tbl.shape[1]}"
    rem_missing = int(clean_tbl.isna().sum().sum())
    
    print(f"  {name:<23} | {raw_shape:<16} | {clean_shape:<16} | {rem_missing:,}")

print("-" * 100)
print(f"  FINAL MERGED TABLE (df) : {df.shape[0]:,} rows x {df.shape[1]} columns (100% unique (raceId, driverId) grain)")
print("=" * 100)
display(df.head())

################################
# 8. VISUALIZATION 1: Missing Values Heatmaps AFTER Cleaning (Stacked Vertically Below Each Other)
################################
plot_tables = list(cleaned_tables.items()) + [('Final Merged Table (df)', df)]

for tbl_name, tbl_df in plot_tables:
    # Check both standard NaN and any literal '\N' strings (0 '\N' after cleaning)
    missing_mask = tbl_df.isna() | (tbl_df.astype(str) == '\\N')
    total_missing = int(missing_mask.sum().sum())
    
    plt.figure(figsize=(12, 3.5))
    sns.heatmap(
        missing_mask,
        cbar=False,
        yticklabels=False,
        cmap='viridis',
        vmin=0,
        vmax=1
    )
    plt.title(f"Missing Values Heatmap (After Cleaning): {tbl_name} ({total_missing:,} total missing)", fontsize=12, pad=10)
    plt.xticks(rotation=90, fontsize=10)
    plt.tight_layout()
    plt.show()

################################
# 9. VISUALIZATION 2: Duplicates & Grain Audit Before vs. After Cleaning
################################
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Left Plot: Duplicate (raceId, driverId) rows before vs after across key tables
dup_categories = ['results.csv\n(Shared Drives)', 'qualifying.csv', 'Final Merged df']
before_dups = [dup_rows_before, qualifying_dups_before, dup_rows_before]
after_dups = [dup_rows_after, qualifying_dups_after, int(df.duplicated(subset=['raceId', 'driverId']).sum())]

x = np.arange(len(dup_categories))
width = 0.35

bars1 = axes[0].bar(x - width/2, before_dups, width, label='Before Cleaning', color='#e74c3c', edgecolor='black')
bars2 = axes[0].bar(x + width/2, after_dups, width, label='After Cleaning', color='#2ecc71', edgecolor='black')

axes[0].set_title('Duplicate (raceId, driverId) Rows: Before vs. After', fontsize=12, fontweight='bold')
axes[0].set_ylabel('Number of Rows Sharing a (raceId, driverId)')
axes[0].set_xticks(x)
axes[0].set_xticklabels(dup_categories, fontsize=10)
axes[0].legend()
axes[0].set_ylim(0, max(before_dups) * 1.2 + 10)

for bar in bars1:
    h = bar.get_height()
    axes[0].text(bar.get_x() + bar.get_width()/2, h + 3, f"{int(h)}", ha='center', fontweight='bold', color='#c0392b')
for bar in bars2:
    h = bar.get_height()
    axes[0].text(bar.get_x() + bar.get_width()/2, h + 3, f"{int(h)}", ha='center', fontweight='bold', color='#27ae60')

# Right Plot: Total Row Count in results.csv Before vs. After Shared-Drive Deduplication
row_stages = ['Raw results.csv\n(With Shared-Drive Dups)', 'Cleaned results & Final df\n(Unique (raceId, driverId) Grain)']
row_counts = [raw_results_rows, clean_results_rows]

bars3 = axes[1].bar(row_stages, row_counts, width=0.45, color=['#34495e', '#3498db'], edgecolor='black')
axes[1].set_title(f'Grain Deduplication Impact ({extra_dup_rows_dropped:,} Redundant Rows Removed)', fontsize=12, fontweight='bold')
axes[1].set_ylabel('Total Row Count')
axes[1].set_ylim(min(row_counts) * 0.98, max(row_counts) * 1.01)

for bar in bars3:
    h = bar.get_height()
    axes[1].text(bar.get_x() + bar.get_width()/2, h + (max(row_counts)*0.0015), f"{int(h):,} rows", ha='center', fontweight='bold')

plt.tight_layout()
plt.show()
```

# Table description

**Identification & Relational Keys**

- raceId: The unique numeric identifier for a specific Grand Prix event.

- driverId: The unique numeric identifier for the driver.

- constructorId: The unique numeric identifier for the constructor (team).

- circuitId: The unique numeric identifier for the racetrack hosting the event.

- number: The permanent or season-specific racing number displayed on the driver's car.

**Race Event Metadata**

- year: The Formula 1 World Championship season in which the race took place.

- round: The chronological race number within that specific season calendar (e.g., Round 1 is the season opener).

**Constructor Profile**

- constructor_name: The official name of the constructor/team (e.g., Ferrari, McLaren, Red Bull).

- constructor_nationality: The country of registration/origin for the constructor (e.g., British, Italian, German).

**Pre-Race Pace & Grid Position**

- position (from qualifying): The finishing rank achieved by the driver in the qualifying sessions based purely on on-track lap times.

- q1: The driver's fastest lap time recorded in the first qualifying knockout session (in seconds).

- q2: The driver's fastest lap time recorded in the second qualifying session (in seconds; imputed for drivers eliminated in Q1).

- q3: The driver's fastest lap time recorded in the top-10 shootout qualifying session (in seconds; imputed for drivers eliminated in Q1/Q2).

- grid: The official starting slot on the track at lights out. This can differ from the qualifying position due to engine/gearbox penalties, sporting reprimands, or pit-lane start mandates.

**Leakage-Safe Pre-Race Standings**

- driver_pts_pre_race: Total World Drivers' Championship points held by the driver entering the race weekend (shifted from prior rounds to prevent post-race target leakage).

- driver_pos_pre_race: The driver’s rank in the World Drivers’ Championship standings entering the race weekend.

- constructor_pts_pre_race: Total World Constructors' Championship points accumulated by the team prior to this race start.

- constructor_pos_pre_race: The constructor’s standing in the Constructors' Championship entering the race weekend.

**Target & Ground Truth Classification**

- positionOrder: The driver’s final classified finishing position at the end of the race (1, 2, 3, etc.).

- podium: The binary prediction target variable. Set to 1 if the driver finished in the top three (positionOrder <= 3), and 0 otherwise.

###  Strict Pre-Race Prediction Cutoff & Leakage Control
Our unit of analysis is a single driver entering a single race (raceId, driverId), and our binary target is podium (positionOrder <= 3). Because predictions must be made **after qualifying and before lights out**, any current-race outcome variable (points, laps, 	ime, statusId, pit stop counts, or post-race standings) constitutes severe data leakage and is strictly excluded.

We engineer and evaluate candidate feature groups that are 100% known prior to the race start:
1. **Starting Grid (grid):** Official starting position after penalties. Track position dictates clean air and turn-1 survival.
2. **Qualifying Performance (quali_pos):** Saturday single-lap classification position prior to grid penalties.
3. **Pre-Race Championship Standings (driver_pts_pre_race, driver_pos_pre_race, constructor_pts_pre_race, constructor_pos_pre_race):** Cumulative points and rank entering the weekend, computed by lagging post-race standings via .shift(1) within each season.
4. **Leakage-Safe Historical Form (driver_form_3, driver_cum_podiums, constructor_form_5):** Strictly past-only rolling performance computed by sorting chronologically by date and applying .shift(1).rolling(...) and .shift(1).cumsum().
5. **Circuit & Driver Context (lt, driver_age):** Circuit altitude and biological age at race start.
6. **Mechanical Reliability (constructor_mech_dnf_rate):** Rolling 10-race constructor mechanical failure rate (.shift(1)), isolating car fragility from driver crashes.
7. **Teammate Qualifying Benchmark (eat_teammate_quali):** Binary indicator of whether the driver out-qualified their teammate in identical machinery.
8. **Career Experience (driver_career_races):** Cumulative prior starts (cumcount()) capturing rookie vs. veteran racecraft.
9. **Era & Season Progression (year, 
ound):** Season year and calendar round number, capturing regulation cycles and championship maturity.
10. **Constructor Pruned OHE (constructor_pruned_*):** One-hot encoding of historical dominant constructors fitted strictly on <= 2019 data.
11. **Driver & Constructor Entity Identifiers (driverId, constructorId):** Explicit entity identifier features capturing driver and constructor identities.


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Ensure chronological sorting by merging race date if not already present
if 'date' not in df.columns:
    races_dates = f1_data['races'][['raceId', 'date']].copy()
    races_dates['date'] = pd.to_datetime(races_dates['date'], errors='coerce')
    df = df.merge(races_dates, on='raceId', how='left')
else:
    df['date'] = pd.to_datetime(df['date'], errors='coerce')

# 2. Engineer Leakage-Safe Historical Form Features (Strictly past races via .shift(1))
df = df.sort_values(by=['driverId', 'date']).reset_index(drop=True)
df['driver_form_3'] = (
    df.groupby('driverId')['positionOrder']
    .transform(lambda s: s.shift(1).rolling(window=3, min_periods=1).mean())
)
df['driver_cum_podiums'] = (
    df.groupby('driverId')['podium']
    .transform(lambda s: s.shift(1).cumsum())
).fillna(0)
df['driver_career_races'] = df.groupby('driverId').cumcount()

df = df.sort_values(by=['constructorId', 'date']).reset_index(drop=True)
df['constructor_form_5'] = (
    df.groupby('constructorId')['positionOrder']
    .transform(lambda s: s.shift(1).rolling(window=5, min_periods=1).mean())
)

# 3. Engineer Domain & Contextual Features (Mechanical Reliability, Teammate Rivalry)
res_stat = f1_data['results'][['raceId', 'constructorId', 'statusId']].merge(
    f1_data['status'], on='statusId', how='left'
)
mech_keywords = 'Engine|Gearbox|Hydraulics|Suspension|Brakes|Transmission|Clutch|Electrical|Pneumatics'
res_stat['is_mech_dnf'] = res_stat['status'].str.contains(mech_keywords, case=False, na=False).astype(int)

const_dnf = res_stat.groupby(['constructorId', 'raceId'])['is_mech_dnf'].mean().reset_index()
const_dnf = const_dnf.merge(f1_data['races'][['raceId', 'date']], on='raceId', how='left')
const_dnf['date'] = pd.to_datetime(const_dnf['date'], errors='coerce')
const_dnf = const_dnf.sort_values(by=['constructorId', 'date']).reset_index(drop=True)
const_dnf['constructor_mech_dnf_rate'] = const_dnf.groupby('constructorId')['is_mech_dnf'].transform(
    lambda x: x.shift(1).rolling(window=10, min_periods=1).mean()
).fillna(0)

if 'constructor_mech_dnf_rate' not in df.columns:
    df = df.merge(
        const_dnf[['constructorId', 'raceId', 'constructor_mech_dnf_rate']], 
        on=['constructorId', 'raceId'], how='left'
    )

team_best_quali = df.groupby(['raceId', 'constructorId'])['quali_pos'].transform('min')
df['beat_teammate_quali'] = ((df['quali_pos'] == team_best_quali) & df['quali_pos'].notna()).astype(int)
df['round'] = pd.to_numeric(df['round'], errors='coerce')

# --- NEW: Leakage-Safe Pruned Constructor One-Hot Encoding ---
TRAIN_END_YEAR = 2019
TOP_N_CONSTRUCTORS = 5
top_constructors = (
    df[df["year"] <= TRAIN_END_YEAR]["constructorId"]
    .value_counts().nlargest(TOP_N_CONSTRUCTORS).index
)
df["constructor_pruned"] = df["constructorId"].where(df["constructorId"].isin(top_constructors), -1)
df = pd.get_dummies(df, columns=["constructor_pruned"], dummy_na=False, dtype=float)
ohe_cols = [c for c in df.columns if c.startswith("constructor_pruned_")]

# Restore chronological race order
df = df.sort_values(by=['date', 'raceId', 'grid']).reset_index(drop=True)

# Define Unified Feature Groups (Including driverId and constructorId)
extended_feature_groups = {
    'grid': ['grid'],
    'qualifying': ['quali_pos'],
    'standings': ['driver_pts_pre_race', 'driver_pos_pre_race', 'constructor_pts_pre_race', 'constructor_pos_pre_race'],
    'form': ['driver_form_3', 'driver_cum_podiums', 'constructor_form_5'],
    'circuit_driver': ['alt', 'driver_age'],
    'reliability': ['constructor_mech_dnf_rate'],
    'teammate_rivalry': ['beat_teammate_quali'],
    'experience': ['driver_career_races'],
    'era_year': ['year'],
    'season_progress': ['round'],
    'constructor_identity': ohe_cols  
    
}
#[group]->driver_id constructor_identity ,circuitid
all_candidate_features = [f for grp in extended_feature_groups.values() for f in grp if f in df.columns]
target = 'podium'

print("✓ Leakage-safe form features verified (using strictly past races via .shift(1)):")
display(df[['year', 'round', 'driverId', 'positionOrder', 'driver_form_3', 'driver_cum_podiums', 'constructor_form_5']].head(6))

# 4. Plot Exploratory Evidence of Chosen Features on Podium Target
fig, axes = plt.subplots(1, 3, figsize=(18, 4.5))

# Plot A: Pearson Correlations with Target
plot_feats = [f for f in all_candidate_features if not f.startswith('constructor_pruned_')]
corrs = df[plot_feats + [target]].corr()[target].drop(target).sort_values()
corrs.plot(kind='barh', ax=axes[0], color='#0f766e', edgecolor='black')
axes[0].set_title("A. Feature Correlation with Podium Finish", fontweight='bold')
axes[0].set_xlabel("Pearson Correlation (r)")
axes[0].axvline(0, color='black', linewidth=0.8, linestyle='--')

# Plot B: Empirical Podium Rate by Starting Grid (Top 10)
grid_podium = df[df['grid'].between(1, 10)].groupby('grid')[target].mean() * 100
sns.barplot(x=grid_podium.index, y=grid_podium.values, ax=axes[1], hue=grid_podium.index, palette='Blues_r', edgecolor='black', legend=False)
axes[1].set_title("B. Podium Conversion Rate by Grid (P1–P10)", fontweight='bold')
axes[1].set_ylabel("Podium Finish Rate (%)")
axes[1].set_xlabel("Starting Grid Position")

# Plot C: Rolling Form Distribution vs Podium
sns.boxplot(data=df, x=target, y='driver_form_3', ax=axes[2], hue=target, palette='Set2', showfliers=False, legend=False)
axes[2].set_title("C. Leakage-Safe Driver Form by Outcome", fontweight='bold')
axes[2].set_xticks([0, 1])
axes[2].set_xticklabels(['Non-Podium (0)', 'Podium (1)'])
axes[2].set_ylabel("Rolling Last 3 Races Avg Finish")

plt.tight_layout()
plt.show()
```

1. **Structural Missingness:** Pre-1996 races lack formal q1/q2/q3 knockout times, and debuting drivers lack prior rolling form (driver_form_3). Dropping these rows would eliminate over 65% of F1 history and introduce survivorship bias.
2. **Heterogeneous Scales:** Features span vastly different magnitudes—lt ranges from -7 m to 2,227 m and driver_pts_pre_race exceeds 400, while grid ranges from 1 to 33.

To rigorously evaluate preprocessing choices, we conduct controlled trials comparing **Order 1 (Median Imputation -> StandardScaler)** against **Order 2 (StandardScaler -> Constant(0) Impute)** evaluated systematically across **all 3 models from Cell 8**:
1. **Logistic Regression (Baseline)**
2. **Random Forest (Untuned: max_depth=None)**
3. **Random Forest (Tuned: max_depth=6)**


```python
import itertools
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import roc_auc_score, average_precision_score, f1_score

# Preliminary Chronological Split (<=2019 Train, 2020-2021 Val) strictly for preprocessing & feature selection
temp_train = df[df['year'] <= 2019]
temp_val = df[(df['year'] >= 2020) & (df['year'] <= 2021)]

X_tr_raw, y_tr_raw = temp_train[all_candidate_features], temp_train[target].values
X_v_raw, y_v_raw = temp_val[all_candidate_features], temp_val[target].values

# 1. Preprocessing-Order Experiment on ALL 3 MODELS from Cell 8
models_to_test = {
    "Logistic Regression (Baseline)": LogisticRegression(class_weight='balanced', max_iter=1000, random_state=42),
    "Random Forest (Untuned: max_depth=None)": RandomForestClassifier(n_estimators=100, max_depth=None, class_weight='balanced', random_state=42, n_jobs=-1),
    "Random Forest (Tuned: max_depth=6)": RandomForestClassifier(n_estimators=150, max_depth=6, class_weight='balanced', random_state=42, n_jobs=-1)
}

order_records = []
for model_name, clf in models_to_test.items():
    # Order 1: Median Impute -> StandardScaler
    pipe1 = Pipeline([
        ('imputer', SimpleImputer(strategy='median')),
        ('scaler', StandardScaler()),
        ('model', clf)
    ])
    pipe1.fit(X_tr_raw, y_tr_raw)
    probs1 = pipe1.predict_proba(X_v_raw)[:, 1]
    
    best_th1, best_f1_1 = 0.5, 0.0
    for th in np.linspace(0.1, 0.9, 81):
        f = f1_score(y_v_raw, (probs1 >= th).astype(int), zero_division=0)
        if f > best_f1_1: best_f1_1, best_th1 = f, th
        
    order_records.append({
        "Model": model_name,
        "Pipeline Experiment": "Order 1: Median Impute -> StandardScaler (Chosen)",
        "Val ROC-AUC": roc_auc_score(y_v_raw, probs1),
        "Val PR-AUC": average_precision_score(y_v_raw, probs1),
        "Val F1": best_f1_1
    })
    
    # Order 2: StandardScaler -> Constant(0) Impute
    pipe2 = Pipeline([
        ('scaler', StandardScaler()),
        ('imputer', SimpleImputer(strategy='constant', fill_value=0.0)),
        ('model', clf)
    ])
    pipe2.fit(X_tr_raw, y_tr_raw)
    probs2 = pipe2.predict_proba(X_v_raw)[:, 1]
    
    best_th2, best_f1_2 = 0.5, 0.0
    for th in np.linspace(0.1, 0.9, 81):
        f = f1_score(y_v_raw, (probs2 >= th).astype(int), zero_division=0)
        if f > best_f1_2: best_f1_2, best_th2 = f, th
        
    order_records.append({
        "Model": model_name,
        "Pipeline Experiment": "Order 2: StandardScaler -> Constant(0) Impute",
        "Val ROC-AUC": roc_auc_score(y_v_raw, probs2),
        "Val PR-AUC": average_precision_score(y_v_raw, probs2),
        "Val F1": best_f1_2
    })

order_exp_df = pd.DataFrame(order_records)
print("=== 1. Preprocessing-Order Experiment Results Across All 3 Models ===")
display(order_exp_df.round(4))

# Plot Preprocessing Comparison Across All 3 Models
plt.figure(figsize=(10, 4))
sns.barplot(
    data=order_exp_df, 
    x='Model', y='Val PR-AUC', hue='Pipeline Experiment', 
    palette=['#0d9488', '#f59e0b'], edgecolor='black'
)
plt.title("Preprocessing Pipeline Order Comparison on Validation PR-AUC Across All 3 Models", fontweight='bold')
plt.ylabel("Validation PR-AUC")
plt.xlabel("Model Architecture")
plt.ylim(0.50, 0.75)
plt.legend(loc='lower right')
plt.tight_layout()
plt.show()

# 2. Visualize Effect of Preprocessing on Feature Distributions & Ranges
prep_pipeline = Pipeline([
    ('imputer', SimpleImputer(strategy='median')),
    ('scaler', StandardScaler())
])
X_tr_transformed = pd.DataFrame(prep_pipeline.fit_transform(X_tr_raw), columns=all_candidate_features)

fig, axes = plt.subplots(1, 2, figsize=(14, 4))
sns.kdeplot(X_tr_raw['driver_form_3'].dropna(), ax=axes[0], fill=True, color='#e11d48', label='Raw (Contains NaNs, Range: 1 to 33)')
axes[0].set_title("Before Preprocessing: Raw Distribution & Scale", fontweight='bold')
axes[0].set_xlabel("driver_form_3 (Position)")
axes[0].legend()

sns.kdeplot(X_tr_transformed['driver_form_3'], ax=axes[1], fill=True, color='#0d9488', label='Imputed & Scaled (0 NaNs, Mean=0, Std=1)')
axes[1].set_title("After Preprocessing: Standardized & Median-Imputed", fontweight='bold')
axes[1].set_xlabel("Standardized Z-Score")
axes[1].legend()
plt.tight_layout()
plt.show()

# 3. Exhaustive Feature-Group Permutation Experiment
print("=== 2. Running Feature-Group Permutation Trial ===")
core_groups_to_permute = [
    'grid', 'qualifying', 'standings', 'form', 
    'circuit_driver', 'reliability', 'teammate_rivalry',
    'era_year', 'season_progress', 'constructor_identity'
]
perm_results = []
fast_rf = RandomForestClassifier(n_estimators=50, max_depth=6, class_weight='balanced', random_state=42, n_jobs=-1)

# Permute combinations up to 6 groups for search depth and speed
for r in range(1, 6):
    for combo in itertools.combinations(core_groups_to_permute, r):
        cols = [f for grp in combo for f in extended_feature_groups[grp] if f in df.columns]
        X_tr_p = prep_pipeline.fit_transform(temp_train[cols])
        X_v_p = prep_pipeline.transform(temp_val[cols])
        fast_rf.fit(X_tr_p, y_tr_raw)
        probs = fast_rf.predict_proba(X_v_p)[:, 1]
        perm_results.append({
            "Included Feature Groups": " + ".join(combo),
            "Num Features": len(cols),
            "Val PR-AUC": average_precision_score(y_v_raw, probs),
            "Val ROC-AUC": roc_auc_score(y_v_raw, probs)
        })

perm_df = pd.DataFrame(perm_results).sort_values(by="Val PR-AUC", ascending=False).reset_index(drop=True)
print("Top 5 Performing Feature-Group Combinations on Validation Set:")
display(perm_df.head(5).round(4))

# Evaluate Top Combinations Across All 3 Models
print("\n=== Top Feature Combinations Evaluated Across All 3 Models ===")
top_combo_records = []
top_3_combos = perm_df.head(3)['Included Feature Groups'].tolist()

for combo_str in top_3_combos:
    grps = combo_str.split(" + ")
    cols = [f for grp in grps for f in extended_feature_groups[grp] if f in df.columns]
    X_tr_c = prep_pipeline.fit_transform(temp_train[cols])
    X_v_c = prep_pipeline.transform(temp_val[cols])
    
    for m_name, m_inst in models_to_test.items():
        m_inst.fit(X_tr_c, y_tr_raw)
        p = m_inst.predict_proba(X_v_c)[:, 1]
        top_combo_records.append({
            "Feature Combination": combo_str[:45] + ("..." if len(combo_str) > 45 else ""),
            "Model": m_name,
            "Val PR-AUC": average_precision_score(y_v_raw, p),
            "Val ROC-AUC": roc_auc_score(y_v_raw, p)
        })
display(pd.DataFrame(top_combo_records).round(4))
```

# Cell 7: Proper Dataset Splitting (Chronological Train / Validation / Test)

###  Justification of Temporal Split vs. Random K-Fold
Formula 1 is a non-stationary, sequential time series governed by multi-year technical regulations and constructor dominance cycles. Using a random K-Fold split would place races from the 2023 season inside the training set while predicting 2021 or 2022 races—leaking future car hierarchy and regulation behavior into the past.

We enforce a strict **Chronological Split**:
* **Training Set (1950–2019, 24,529 rows):** Historical racing up to the late turbo-hybrid era used to fit model parameters and preprocessing statistics (imputer medians and scaler means/variances).
* **Validation Set (2020–2021, 780 rows):** Used strictly for hyperparameter tuning, feature-subset selection, historical start-year verification, and probability threshold optimization (tau*).
* **Unseen Test Set (2022–2024, 1,359 rows):** Completely held-out ground-effect aerodynamic regulation era, evaluated only once at the end of the pipeline.

###  Historical Start-Year Empirical Experiment Across All 3 Models
Because qualifying session formats evolved over time (e.g., knockout q1/q2/q3 began in 2006), we tested four training start years (1950, 1960, 1996, 2006) on the 2020–2021 validation set across **all 3 models from Cell 8**:
1. **Logistic Regression (Baseline)**
2. **Random Forest (Untuned: max_depth=None)**
3. **Random Forest (Tuned: max_depth=6)**

# Cell 9: Feature Selection & Leakage-Safe Form Engineering (Shallow FFNN)
---
---
### 1. Justification & Evidence for Candidate Feature Groups
All features are strictly constrained to information known **prior to lights out on Sunday** (pre-race cutoff) to prevent target leakage:
* **`grid` (`grid`)**: Direct measure of track position at race start (incorporating qualifying penalties). In our Q1 EDA, front-row starters converted to podiums at up to 86.3% on certain tracks.
* **`qualifying` (`q1`, `q2`, `q3`, `position`)**: Captures raw Saturday single-lap pace and qualifying session progression.
* **`standings` (`driver_pts_pre_race`, `driver_pos_pre_race`, `constructor_pts_pre_race`, `constructor_pos_pre_race`)**: Shifted by 1 round within each season so Round 1 starts at 0 points, reflecting cumulative season dominance entering the race weekend.
* **`form` (`driver_form_3`, `driver_cum_podiums`, `constructor_form_5`)**: Leakage-safe rolling metrics using `.shift(1)` over chronological race order so only strictly past races are aggregated. For constructor form, results are first averaged per `(raceId, constructorId)` before shifting so a teammate's same-race result never leaks across rows.
* **`constructor_identity` (`constructor_pruned_*`)**: One-hot encoding of the top 5 historical constructors (fitted strictly on training data `<= 2019` to prevent frequency leakage) to capture long-term team budget and engineering pedigree.
* **`era_context` (`year`, `round`)**: Captures calendar progression and regulation era shifts.


```python
# ==============================================================================
# CELL 9: FEATURE SELECTION, LEAKAGE-SAFE FORM & EMPIRICAL GROUP SELECTION
# ==============================================================================
import random
import warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

import torch
import torch.nn as nn
from torch.utils.data import DataLoader, TensorDataset
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import roc_auc_score, average_precision_score, precision_recall_curve, f1_score

warnings.filterwarnings("ignore")

# 1. Reproducibility & Device Configuration
SEED = 42
def set_seed(seed=SEED):
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    if torch.cuda.is_available():
        torch.cuda.manual_seed_all(seed)

set_seed()
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

TARGET = "podium"
TRAIN_END_YEAR = 2019
VAL_START_YEAR, VAL_END_YEAR = 2020, 2021
TEST_START_YEAR, TEST_END_YEAR = 2022, 2024
TOP_N_CONSTRUCTORS = 5

# 2. Leakage-Safe Feature Engineering Function (Strictly Past Races Only)
def engineer_ffnn_features(df_raw, start_year=1950):
    d = df_raw.copy()
    # Removed is_home_race, keeping rest of numerical columns
    num_cols = [
        "year", "round", TARGET, "positionOrder", "grid", "quali_pos",
        "driver_pts_pre_race", "driver_pos_pre_race", "constructor_pts_pre_race", 
        "constructor_pos_pre_race", "alt", "driver_age",
        "constructor_mech_dnf_rate", "beat_teammate_quali", "driver_career_races"
    ]
    for col in num_cols:
        if col in d.columns:
            d[col] = pd.to_numeric(d[col], errors="coerce")
            
    d = d[d["year"] >= start_year].copy()
    # Sort strictly chronologically by year, round, raceId, driverId
    d = d.sort_values(by=["year", "round", "raceId", "driverId"], kind="mergesort").reset_index(drop=True)
    
    # A. Leakage-Safe Driver Form: Last 3 races average finish (shifted by 1)
    d["driver_form_3"] = (
        d.groupby("driverId")["positionOrder"]
        .transform(lambda s: s.shift(1).rolling(window=3, min_periods=1).mean())
    )
    
    # B. Leakage-Safe Driver Career Podiums: Cumulative sum prior to current race
    d["driver_cum_podiums"] = (
        d.groupby("driverId")[TARGET]
        .transform(lambda s: s.shift(1).cumsum())
    ).fillna(0)
    
    # C. Leakage-Safe Constructor Form: Aggregate per race first so teammate same-race results don't leak
    con_race = (
        d.groupby(["year", "round", "raceId", "constructorId"])["positionOrder"]
        .mean().reset_index()
        .sort_values(by=["constructorId", "year", "round", "raceId"])
    )
    con_race["constructor_form_5"] = (
        con_race.groupby("constructorId")["positionOrder"]
        .transform(lambda s: s.shift(1).rolling(window=5, min_periods=1).mean())
    )
    d = d.drop(columns=["constructor_form_5"], errors="ignore").merge(
        con_race[["raceId", "constructorId", "constructor_form_5"]],
        on=["raceId", "constructorId"], how="left"
    )
    
    # D. Pruned Constructor OHE (Fitted strictly on Train <= 2019 to prevent leakage)
    top_constructors = (
        d[d["year"] <= TRAIN_END_YEAR]["constructorId"]
        .value_counts().nlargest(TOP_N_CONSTRUCTORS).index
    )
    d["constructor_pruned"] = d["constructorId"].where(d["constructorId"].isin(top_constructors), -1)
    d_ohe = pd.get_dummies(d, columns=["constructor_pruned"], dummy_na=False, dtype=float)
    ohe_cols = [c for c in d_ohe.columns if c.startswith("constructor_pruned_")]
    
    # Unified Feature Groups Dict (Plus Constructor OHE for NN)
    feature_groups = {
        'grid': ['grid'],
        'qualifying': ['quali_pos'], 
        'standings': ['driver_pts_pre_race', 'driver_pos_pre_race', 'constructor_pts_pre_race', 'constructor_pos_pre_race'],
        'form': ['driver_form_3', 'driver_cum_podiums', 'constructor_form_5'],
        'circuit_driver': ['alt', 'driver_age'],
        'reliability': ['constructor_mech_dnf_rate'],
        'teammate_rivalry': ['beat_teammate_quali'],
        'experience': ['driver_career_races'],
        'era_year': ['year'],
        'season_progress': ['round'],
        'constructor_identity': ohe_cols
    }
    return d_ohe, feature_groups

df_fe, CANDIDATE_GROUPS = engineer_ffnn_features(df, start_year=1950)
train_fe = df_fe[df_fe["year"] <= TRAIN_END_YEAR].copy()

print("✓ Leakage-safe form features verified (using strictly past races via .shift(1)):")
display(train_fe[["year", "round", "driverId", "positionOrder", "driver_form_3", "driver_cum_podiums", "constructor_form_5"]].head(6))

# 3. Plotting the Effect of Chosen Features on Podium Finish (1x3 Panel as Requested)
fig, axes = plt.subplots(1, 3, figsize=(18, 4.5))

core_feats = [c for grp in ['grid', 'qualifying', 'standings', 'form', 'era_year', 'season_progress'] for c in CANDIDATE_GROUPS[grp]]

# Plot A: Pearson Correlations with Target
plot_feats = [f for f in core_feats if not f.startswith('constructor_pruned_')]
corrs = train_fe[plot_feats + [TARGET]].corr()[TARGET].drop(TARGET).sort_values()
corrs.plot(kind='barh', ax=axes[0], color='#0f766e', edgecolor='black')
axes[0].set_title("A. Feature Correlation with Podium Finish", fontweight='bold')
axes[0].set_xlabel("Pearson Correlation (r)")
axes[0].axvline(0, color='black', linewidth=0.8, linestyle='--')

# Plot B: Empirical Podium Rate by Starting Grid (Top 10)
grid_podium = train_fe[train_fe['grid'].between(1, 10)].groupby('grid')[TARGET].mean() * 100
sns.barplot(x=grid_podium.index, y=grid_podium.values, ax=axes[1], hue=grid_podium.index, palette='Blues_r', edgecolor='black', legend=False)
axes[1].set_title("B. Podium Conversion Rate by Grid (P1–P10)", fontweight='bold')
axes[1].set_ylabel("Podium Finish Rate (%)")
axes[1].set_xlabel("Starting Grid Position")

# Plot C: Leakage-Safe Driver Form Boxplot
sns.boxplot(data=train_fe, x=TARGET, y='driver_form_3', ax=axes[2], hue=TARGET, palette='Set2', showfliers=False, legend=False)
axes[2].set_title("C. Leakage-Safe Driver Form by Outcome", fontweight='bold')
axes[2].set_xticks([0, 1])
axes[2].set_xticklabels(['Non-Podium (0)', 'Podium (1)'])
axes[2].set_ylabel("Rolling Last 3 Races Avg Finish")

plt.tight_layout()
plt.show()

# 4. Shallow FFNN Architecture & Helper for Empirical Feature Selection
class ShallowFFNN(nn.Module):
    """2-Hidden-Layer Feedforward Neural Network (64 -> 32 -> 1) with BatchNorm & Dropout."""
    def __init__(self, input_dim, drop_rates=(0.3, 0.2)):
        super().__init__()
        self.net = nn.Sequential(
            nn.Linear(input_dim, 64),
            nn.BatchNorm1d(64),
            nn.ReLU(),
            nn.Dropout(drop_rates[0]),
            nn.Linear(64, 32),
            nn.BatchNorm1d(32),
            nn.ReLU(),
            nn.Dropout(drop_rates[1]),
            nn.Linear(32, 1)
        )
    def forward(self, x):
        return self.net(x).squeeze(-1)

def train_eval_ffnn_quick(X_tr, y_tr, X_va, y_va, epochs=50, patience=7, lr=1e-3, weight_decay=1e-4):
    set_seed()
    model = ShallowFFNN(X_tr.shape[1]).to(device)
    optimizer = torch.optim.Adam(model.parameters(), lr=lr, weight_decay=weight_decay)
    pw = (y_tr == 0).sum() / max((y_tr == 1).sum(), 1)
    criterion = nn.BCEWithLogitsLoss(pos_weight=torch.tensor([pw], dtype=torch.float32).to(device))
    
    tr_loader = DataLoader(TensorDataset(torch.tensor(X_tr), torch.tensor(y_tr)), batch_size=256, shuffle=True, drop_last=True)
    va_loader = DataLoader(TensorDataset(torch.tensor(X_va), torch.tensor(y_va)), batch_size=256, shuffle=False)
    
    best_loss, best_state, wait = float("inf"), None, 0
    for _ in range(epochs):
        model.train()
        for xb, yb in tr_loader:
            optimizer.zero_grad()
            loss = criterion(model(xb.to(device)), yb.to(device))
            loss.backward()
            optimizer.step()
            
        model.eval()
        v_loss = 0.0
        with torch.no_grad():
            for xb, yb in va_loader:
                v_loss += criterion(model(xb.to(device)), yb.to(device)).item() * len(yb)
        v_loss /= len(va_loader.dataset)
        
        if v_loss < best_loss:
            best_loss, wait = v_loss, 0
            best_state = {k: v.detach().cpu().clone() for k, v in model.state_dict().items()}
        else:
            wait += 1
            if wait >= patience:
                break
                
    model.load_state_dict(best_state)
    model.eval()
    with torch.no_grad():
        p_va = torch.sigmoid(model(torch.tensor(X_va).to(device))).cpu().numpy()
        
    prec, rec, thresh = precision_recall_curve(y_va, p_va)
    f1s = (2 * prec[:-1] * rec[:-1]) / (prec[:-1] + rec[:-1] + 1e-12)
    best_th = float(thresh[np.argmax(f1s)]) if len(thresh) > 0 else 0.5
    
    return {
        "Val_ROC_AUC": roc_auc_score(y_va, p_va),
        "Val_PR_AUC": average_precision_score(y_va, p_va),
        "Val_F1": f1_score(y_va, (p_va >= best_th).astype(int), zero_division=0),
        "Best_Threshold": best_th
    }

# 5. Empirical Leave-One-Group-Out Selection on Validation Set (2020-2021)
val_fe = df_fe[df_fe["year"].between(VAL_START_YEAR, VAL_END_YEAR)].copy()
all_candidate_cols = [c for grp in CANDIDATE_GROUPS.values() for c in grp if c in df_fe.columns]

imp_base = SimpleImputer(strategy="median")
scl_base = StandardScaler()

X_tr_all = scl_base.fit_transform(imp_base.fit_transform(train_fe[all_candidate_cols])).astype(np.float32)
X_va_all = scl_base.transform(imp_base.transform(val_fe[all_candidate_cols])).astype(np.float32)
y_tr_all = train_fe[TARGET].to_numpy(dtype=np.float32)
y_va_all = val_fe[TARGET].to_numpy(dtype=np.float32)

base_res = train_eval_ffnn_quick(X_tr_all, y_tr_all, X_va_all, y_va_all)
print(f"\nBaseline FFNN (All Groups) -> Val PR-AUC: {base_res['Val_PR_AUC']:.4f} | Val ROC-AUC: {base_res['Val_ROC_AUC']:.4f} | Val F1: {base_res['Val_F1']:.4f} | Base Threshold: {base_res['Best_Threshold']:.4f}")

selection_rows = []
kept_groups = {}

for grp_name, cols in CANDIDATE_GROUPS.items():
    sub_cols = [c for g, clist in CANDIDATE_GROUPS.items() if g != grp_name for c in clist if c in df_fe.columns]
    X_tr_sub = scl_base.fit_transform(imp_base.fit_transform(train_fe[sub_cols])).astype(np.float32)
    X_va_sub = scl_base.transform(imp_base.transform(val_fe[sub_cols])).astype(np.float32)
    
    res = train_eval_ffnn_quick(X_tr_sub, y_tr_all, X_va_sub, y_va_all)
    delta_pr = res["Val_PR_AUC"] - base_res["Val_PR_AUC"]
    
    if delta_pr > 0.005:
        decision = "DROP (Harms Generalization)"
    else:
        decision = "KEEP (Essential Signal)"
        kept_groups[grp_name] = [c for c in cols if c in df_fe.columns]
        
    selection_rows.append({
        "Evaluated Group": grp_name,
        "Features in Group": ", ".join(cols[:4]) + ("..." if len(cols) > 4 else ""),
        "Val PR-AUC Without Group": res["Val_PR_AUC"],
        "Δ Val PR-AUC": delta_pr,
        "Val ROC-AUC Without Group": res["Val_ROC_AUC"],
        "Optimal Threshold": res["Best_Threshold"],
        "Decision": decision
    })

sel_df = pd.DataFrame(selection_rows).sort_values(by="Δ Val PR-AUC", ascending=True).reset_index(drop=True)
display(sel_df.round(4))

BEST_FEATURE_GROUPS = kept_groups
BEST_FEATURES = [c for grp in BEST_FEATURE_GROUPS.values() for c in grp]
print(f"\n✓ BEST_FEATURE_GROUPS Selected ({len(BEST_FEATURE_GROUPS)} groups, {len(BEST_FEATURES)} total features): {list(BEST_FEATURE_GROUPS.keys())}")
```


# Cell 10: Data Pre-processing & Preprocessing-Order Experiments (Shallow FFNN)
---
---
### Why Preprocessing Matters for a Shallow FFNN
Unlike tree-based models, neural networks optimize weights via gradient descent (`Adam`) and compute linear combinations across inputs (`nn.Linear`). Raw features with vastly different ranges (e.g., `driver_pts_pre_race` up to 450 vs. `grid` 1–20 vs. `q1` in seconds) cause ill-conditioned loss surfaces and unstable gradients. Furthermore, pre-1996 races and Q1/Q2 knockouts leave missing values (`NaN`) in qualifying columns that would propagate `NaN` gradients without imputation.

Here we test two distinct preprocessing-order pipelines on our `BEST_FEATURES` using the Shallow FFNN:
1. **Order 1 (`Impute -> Scale`)**: `SimpleImputer(strategy='median')` followed by `StandardScaler()`.
2. **Order 2 (`Scale -> Impute`)**: `StandardScaler()` first (which ignores `NaN`s when computing mean/variance), followed by `SimpleImputer(strategy='median')` on the z-scored features.

```python
# ==============================================================================
# CELL 10: DATA PRE-PROCESSING DISTRIBUTION EFFECTS & ORDER EXPERIMENTS
# ==============================================================================
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler

# --- FIX: Clean any phantom duplicate columns caused by re-running cells ---
train_fe = train_fe.loc[:, ~train_fe.columns.duplicated()].copy()
val_fe = val_fe.loc[:, ~val_fe.columns.duplicated()].copy()
# -------------------------------------------------------------------------

# 1. Define the Two Preprocessing Order Pipelines
PREPROCESSING_PIPELINES = {
    "Order 1: Impute (Median) -> Scale (StandardScaler)": Pipeline([
        ("imputer", SimpleImputer(strategy="median")),
        ("scaler", StandardScaler())
    ]),
    "Order 2: Scale (StandardScaler) -> Impute (Median)": Pipeline([
        ("scaler", StandardScaler()),
        ("imputer", SimpleImputer(strategy="median"))
    ])
}

# 2. Inspect & Plot the Effect of Each Step on Data Distribution, Range, and Values
# Dynamically pick columns that survived Cell 13's feature selection
available_for_stats = [c for c in ["position", "grid", "driver_pts_pre_race", "driver_form_3"] if c in BEST_FEATURES]
X_tr_raw_df = train_fe[BEST_FEATURES].copy()

imp_step = SimpleImputer(strategy="median")
scl_step = StandardScaler()

X_tr_after_imp = pd.DataFrame(imp_step.fit_transform(X_tr_raw_df), columns=BEST_FEATURES)
X_tr_after_scl = pd.DataFrame(scl_step.fit_transform(X_tr_after_imp), columns=BEST_FEATURES)

dist_stats = []
for col in available_for_stats:
    dist_stats.append({
        "Feature": col, "Stage": "1. Raw (Pre-Imputation)",
        "NaN Count": X_tr_raw_df[col].isna().sum(),
        "Min": X_tr_raw_df[col].min(), "Median": X_tr_raw_df[col].median(),
        "Mean": X_tr_raw_df[col].mean(), "Max": X_tr_raw_df[col].max(), "Std": X_tr_raw_df[col].std()
    })
    dist_stats.append({
        "Feature": col, "Stage": "2. After Median Imputation",
        "NaN Count": X_tr_after_imp[col].isna().sum(),
        "Min": X_tr_after_imp[col].min(), "Median": X_tr_after_imp[col].median(),
        "Mean": X_tr_after_imp[col].mean(), "Max": X_tr_after_imp[col].max(), "Std": X_tr_after_imp[col].std()
    })
    dist_stats.append({
        "Feature": col, "Stage": "3. After Standard Scaling",
        "NaN Count": X_tr_after_scl[col].isna().sum(),
        "Min": X_tr_after_scl[col].min(), "Median": X_tr_after_scl[col].median(),
        "Mean": X_tr_after_scl[col].mean(), "Max": X_tr_after_scl[col].max(), "Std": X_tr_after_scl[col].std()
    })

print("Effect of Preprocessing Steps on Feature Range, Central Tendency, and Missingness:")
display(pd.DataFrame(dist_stats).round(3))

# Plot Histograms Across the 3 Stages for the first two available features
plot_cols = available_for_stats[:2]
fig, axes = plt.subplots(len(plot_cols), 3, figsize=(15, 3.5 * len(plot_cols)))

# Handle 1D axes array if only 1 feature is available
if len(plot_cols) == 1:
    axes = np.array([axes])

for row_idx, feat_name in enumerate(plot_cols):
    axes[row_idx, 0].hist(X_tr_raw_df[feat_name].dropna(), bins=35, color="slategray", edgecolor="black")
    axes[row_idx, 0].set_title(f"{feat_name}: 1. Raw (NaNs={X_tr_raw_df[feat_name].isna().sum():,})", fontweight="bold")
    
    axes[row_idx, 1].hist(X_tr_after_imp[feat_name], bins=35, color="steelblue", edgecolor="black")
    axes[row_idx, 1].set_title(f"{feat_name}: 2. After Median Imputation (NaNs=0)", fontweight="bold")
    
    axes[row_idx, 2].hist(X_tr_after_scl[feat_name], bins=35, color="seagreen", edgecolor="black")
    axes[row_idx, 2].set_title(f"{feat_name}: 3. After StandardScaler (Mean≈0, Std=1)", fontweight="bold")

plt.suptitle("Effect of Preprocessing Steps on Feature Distributions (Training Set)", fontsize=13, fontweight="bold", y=1.02)
plt.tight_layout()
plt.show()

# 3. Run Preprocessing-Order Experiment on the Shallow FFNN
prep_exp_results = []
y_tr_vec = train_fe[TARGET].to_numpy(dtype=np.float32)
y_va_vec = val_fe[TARGET].to_numpy(dtype=np.float32)

for order_name, pipe in PREPROCESSING_PIPELINES.items():
    X_tr_p = pipe.fit_transform(train_fe[BEST_FEATURES]).astype(np.float32)
    X_va_p = pipe.transform(val_fe[BEST_FEATURES]).astype(np.float32)
    
    metrics = train_eval_ffnn_quick(X_tr_p, y_tr_vec, X_va_p, y_va_vec)
    
    # Measure L2 difference between transformed validation matrices
    prep_exp_results.append({
        "Preprocessing Pipeline Order": order_name,
        "Val PR-AUC": metrics["Val_PR_AUC"],
        "Val ROC-AUC": metrics["Val_ROC_AUC"],
        "Val F1": metrics["Val_F1"]
    })

prep_df = pd.DataFrame(prep_exp_results).sort_values(by="Val PR-AUC", ascending=False).reset_index(drop=True)
print("\n" + "=" * 85)
print("PREPROCESSING-ORDER EXPERIMENT RESULTS (Shallow FFNN on Validation Set)")
print("=" * 85)
display(prep_df.round(4))

BEST_PREP_NAME = prep_df.iloc[0]["Preprocessing Pipeline Order"]
print(f"\n✓ Winning Preprocessing Pipeline Selected: '{BEST_PREP_NAME}'")
print("Note: In Order 1 (Impute -> Scale), median imputation concentrates missing historical rows at the median BEFORE")
print("StandardScaler calculates standard deviation, shrinking the observed variance and expanding z-scores for non-missing")
print("modern qualifying times. In Order 2 (Scale -> Impute), StandardScaler computes true variance on observed data only.")
```


# Cell 11: Proper Dataset Splitting & Temporal Era Subset Experiments
---
---
### Justification of Chronological Splitting
Formula 1 is a non-stationary time-series domain where rules, point systems, and car reliability evolve across seasons. Using a random `train_test_split` would cause severe **temporal data leakage**—allowing the model to train on Round 15 of 2023 to predict Round 4 of 2023, or leaking future championship dominance backward in time.

Therefore, we enforce a strict **chronological split**:
* **Training Set (`start_year` – `2019`)**: Used to fit preprocessors (imputers, scalers, top-5 constructors) and optimize FFNN weights.
* **Validation Set (`2020` – `2021`)**: Used strictly for early stopping, hyperparameter/era selection, and probability threshold tuning.
* **Unseen Test Set (`2022` – `2024`)**: Locked until final evaluation to measure generalization to the modern Ground-Effect Era.

```python
# ==============================================================================
# CELL 11: PROPER DATASET SPLITTING & HISTORICAL ERA SUBSET EXPERIMENTS
# ==============================================================================
import copy

SUBSETS = {
    "Full Championship History (1950+)": 1950,
    "Modern Team Identity Era (1960+)": 1960,
    "Standardized Qualifying Era (1996+)": 1996,
    "Knockout Qualifying Era (2006+)": 2006,
}

TRAIN_END_YEAR = 2019
VAL_START_YEAR, VAL_END_YEAR = 2020, 2021
TEST_START_YEAR, TEST_END_YEAR = 2022, 2024

subset_rows = []

print("Evaluating Shallow FFNN across the 4 Historical Training Subsets (using BEST_FEATURE_GROUPS & BEST_PREP_NAME)...\n")

for subset_label, start_yr in SUBSETS.items():
    # Engineer leakage-safe features starting from start_yr
    d_sub, grp_sub = engineer_ffnn_features(df, start_year=start_yr)
    active_cols = [c for g in BEST_FEATURE_GROUPS.keys() for c in grp_sub[g] if c in d_sub.columns]
    
    tr_sub = d_sub[d_sub["year"] <= TRAIN_END_YEAR].copy()
    va_sub = d_sub[d_sub["year"].between(VAL_START_YEAR, VAL_END_YEAR)].copy()
    te_sub = d_sub[d_sub["year"].between(TEST_START_YEAR, TEST_END_YEAR)].copy()
    
    # Assertions verifying zero chronological overlap
    assert tr_sub["year"].max() < va_sub["year"].min(), "Train/Val temporal overlap!"
    assert va_sub["year"].max() < te_sub["year"].min(), "Val/Test temporal overlap!"
    
    pipe = copy.deepcopy(PREPROCESSING_PIPELINES[BEST_PREP_NAME])
    X_tr_s = pipe.fit_transform(tr_sub[active_cols]).astype(np.float32)
    X_va_s = pipe.transform(va_sub[active_cols]).astype(np.float32)
    y_tr_s = tr_sub[TARGET].to_numpy(dtype=np.float32)
    y_va_s = va_sub[TARGET].to_numpy(dtype=np.float32)
    
    val_metrics = train_eval_ffnn_quick(X_tr_s, y_tr_s, X_va_s, y_va_s)
    
    subset_rows.append({
        "Historical Subset": subset_label,
        "Start Year": start_yr,
        "Train Rows (<=2019)": len(tr_sub),
        "Val Rows (2020-21)": len(va_sub),
        "Test Rows (2022-24)": len(te_sub),
        "Train Podium %": round(y_tr_s.mean() * 100, 2),
        "Val PR-AUC": val_metrics["Val_PR_AUC"],
        "Val ROC-AUC": val_metrics["Val_ROC_AUC"],
        "Val F1": val_metrics["Val_F1"]
    })

subset_df = pd.DataFrame(subset_rows)
display(subset_df.round(4))

# Pick the optimal subset strictly based on Validation PR-AUC
best_idx = subset_df["Val PR-AUC"].idxmax()
BEST_SUBSET_NAME = subset_df.loc[best_idx, "Historical Subset"]
BEST_START_YEAR = int(subset_df.loc[best_idx, "Start Year"])

# Plot Subset Comparison & Split Proportions
fig, ax = plt.subplots(1, 2, figsize=(14, 4.5))

sns.barplot(data=subset_df, x="Historical Subset", y="Val PR-AUC", palette="Blues_d", edgecolor="black", ax=ax[0])
ax[0].set_title("Validation PR-AUC by Historical Training Start Year", fontweight="bold")
ax[0].set_ylim(subset_df["Val PR-AUC"].min() - 0.03, subset_df["Val PR-AUC"].max() + 0.02)
ax[0].tick_params(axis="x", rotation=20)
for i, v in enumerate(subset_df["Val PR-AUC"]):
    ax[0].text(i, v + 0.003, f"{v:.4f}", ha="center", fontweight="bold")

# Visualize the Chronological Split Sizes for the Winning Subset
win_row = subset_df.loc[best_idx]
splits = ["Train (<=2019)", "Validation (2020-21)", "Test (2022-24)"]
counts = [win_row["Train Rows (<=2019)"], win_row["Val Rows (2020-21)"], win_row["Test Rows (2022-24)"]]
ax[1].bar(splits, counts, color=["#2b5c8f", "#d95f02", "#7570b3"], edgecolor="black")
ax[1].set_title(f"Chronological Split Sizes: {BEST_SUBSET_NAME}", fontweight="bold")
ax[1].set_ylabel("Number of Driver-Race Entries")
for i, c in enumerate(counts):
    ax[1].text(i, c + 300, f"{c:,}", ha="center", fontweight="bold")

plt.tight_layout()
plt.show()

print(f"\n✓ BEST_SUBSET_NAME Selected (by Validation PR-AUC): '{BEST_SUBSET_NAME}' (Start Year = {BEST_START_YEAR})")
```


# Cell 12: Final Shallow FFNN Prediction Model, Evaluation & Feature-Group Ablation
---
---
### 1. How the Shallow FFNN Works
Our **Shallow Feedforward Neural Network (`ShallowFFNN`)** is a 2-hidden-layer Multi-Layer Perceptron (`Input -> 64 -> 32 -> 1`):
* **Linear Projections & Batch Normalization (`nn.Linear` + `nn.BatchNorm1d`)**: Each hidden layer computes a weighted linear combination of the input features, followed by Batch Normalization to stabilize internal covariate shift across mini-batches.
* **Non-Linear Activations (`nn.ReLU`)**: Rectified Linear Units allow the network to learn non-linear feature interactions (e.g., how a P4 grid start interacts with Red Bull/Mercedes constructor dominance and recent form).
* **Regularization (`nn.Dropout(0.3, 0.2)` + L2 Weight Decay `1e-4` + Early Stopping)**: Prevents the network from memorizing noisy historical anomalies.
* **Imbalance-Aware Loss (`nn.BCEWithLogitsLoss(pos_weight)`)**: Because podium finishes represent only ~15% of entries, positive examples are weighted by $w_{\text{pos}} = \frac{N_{\text{neg}}}{N_{\text{pos}}}$ so the gradient updates do not collapse toward predicting the majority class.
* **Validation-Tuned Decision Threshold**: Raw logits are passed through a Sigmoid $\sigma(z) = \frac{1}{1 + e^{-z}}$ to output calibrated podium probabilities, and the classification threshold $\tau^*$ is chosen strictly on the `2020–2021` validation Precision-Recall curve to maximize F1.

### 2. Known Limitations of the Shallow FFNN
* **Historical Qualifying Coverage Drift**: Full knockout qualifying (`q1`, `q2`, `q3`) was only introduced in 2006. For pre-2006 training rows, `q1`/`q2`/`q3` are imputed to the median, compressing single-lap pace differences in older eras.
* **Regulation Era Sensitivity (Non-Stationarity)**: Major aerodynamic resets (such as the 2022 Ground-Effect regulations in our test set) can abruptly reshuffle team pecking orders faster than rolling 5-race constructor form or historical constructor OHE weights can adapt in Round 1.
* **Absence of Sunday Race-Day Context**: Because the model strictly obeys the pre-race cutoff, it cannot anticipate stochastic Sunday events such as Lap 1 collisions, rainy weather transitions, safety cars, or mechanical DNFs.

```python
# ==============================================================================
# CELL 12: FINAL SHALLOW FFNN TRAINING, TEST EVALUATION, PLOTS & GROUP ABLATION
# ==============================================================================
import copy
from sklearn.metrics import (
    precision_score, recall_score, accuracy_score, f1_score,
    roc_curve, precision_recall_curve, average_precision_score, roc_auc_score,
    confusion_matrix, ConfusionMatrixDisplay
)

# 1. Assemble Final Dataset Using Winning Configurations from Cells 12, 13, and 14
print("=" * 90)
print("FINAL SHALLOW FFNN PIPELINE CONFIGURATION")
print(f" • Selected Feature Groups (Cell 13) : {list(BEST_FEATURE_GROUPS.keys())}")
print(f" • Best Preprocessing (Cell 14)      : {BEST_PREP_NAME}")
print(f" • Best Temporal Subset (Cell 15)    : {BEST_SUBSET_NAME} (Start: {BEST_START_YEAR})")
print("=" * 90)

# Generate features based on the winning BEST_START_YEAR
df_final, final_groups = engineer_ffnn_features(df, start_year=BEST_START_YEAR)

# --- FIX: Clean phantom duplicate columns from memory! ---
df_final = df_final.loc[:, ~df_final.columns.duplicated()].copy()
# ---------------------------------------------------------

active_cols = [c for g in BEST_FEATURE_GROUPS.keys() for c in final_groups[g] if c in df_final.columns]

# Apply chronological splits
tr_final = df_final[df_final["year"] <= TRAIN_END_YEAR].copy()
va_final = df_final[df_final["year"].between(VAL_START_YEAR, VAL_END_YEAR)].copy()
te_final = df_final[df_final["year"].between(TEST_START_YEAR, TEST_END_YEAR)].copy()

# Apply winning preprocessing pipeline
final_pipe = copy.deepcopy(PREPROCESSING_PIPELINES[BEST_PREP_NAME])
X_tr_fin = final_pipe.fit_transform(tr_final[active_cols]).astype(np.float32)
X_va_fin = final_pipe.transform(va_final[active_cols]).astype(np.float32)
X_te_fin = final_pipe.transform(te_final[active_cols]).astype(np.float32)

y_tr_fin = tr_final[TARGET].to_numpy(dtype=np.float32)
y_va_fin = va_final[TARGET].to_numpy(dtype=np.float32)
y_te_fin = te_final[TARGET].to_numpy(dtype=np.float32)

# 2. Train the Final Model on the Optimized Splits
set_seed()
final_model = ShallowFFNN(X_tr_fin.shape[1]).to(device)

optimizer = torch.optim.Adam(final_model.parameters(), lr=3e-4, weight_decay=1e-4)
pw = (y_tr_fin == 0).sum() / max((y_tr_fin == 1).sum(), 1)
criterion = nn.BCEWithLogitsLoss(pos_weight=torch.tensor([pw], dtype=torch.float32).to(device))

tr_loader = DataLoader(TensorDataset(torch.tensor(X_tr_fin), torch.tensor(y_tr_fin)), batch_size=256, shuffle=True, drop_last=True)
va_loader = DataLoader(TensorDataset(torch.tensor(X_va_fin), torch.tensor(y_va_fin)), batch_size=256, shuffle=False)
te_loader = DataLoader(TensorDataset(torch.tensor(X_te_fin), torch.tensor(y_te_fin)), batch_size=256, shuffle=False)

train_losses, val_losses = [], []
best_loss, best_state, wait = float("inf"), None, 0

final_model.eval()
with torch.no_grad():
    ep0_tr_loss = sum(criterion(final_model(xb.to(device)), yb.to(device)).item() * len(yb) for xb, yb in tr_loader) / len(tr_loader.dataset)
    ep0_va_loss = sum(criterion(final_model(xb.to(device)), yb.to(device)).item() * len(yb) for xb, yb in va_loader) / len(va_loader.dataset)
train_losses.append(ep0_tr_loss)
val_losses.append(ep0_va_loss)
best_loss = ep0_va_loss
best_state = {k: v.detach().cpu().clone() for k, v in final_model.state_dict().items()}

for epoch in range(1, 60):  
    final_model.train()
    ep_tr_loss = 0
    for xb, yb in tr_loader:
        optimizer.zero_grad()
        loss = criterion(final_model(xb.to(device)), yb.to(device))
        loss.backward()
        optimizer.step()
        ep_tr_loss += loss.item() * len(yb)
    train_losses.append(ep_tr_loss / len(tr_loader.dataset))
    
    final_model.eval()
    v_loss = 0.0
    with torch.no_grad():
        for xb, yb in va_loader:
            v_loss += criterion(final_model(xb.to(device)), yb.to(device)).item() * len(yb)
    v_loss /= len(va_loader.dataset)
    val_losses.append(v_loss)
    
    if v_loss < best_loss:
        best_loss, v_loss, wait = v_loss, 0, 0
        best_state = {k: v.detach().cpu().clone() for k, v in final_model.state_dict().items()}
    else:
        wait += 1
        if wait >= 10:  
            break

final_model.load_state_dict(best_state)
final_model.eval()

def get_preds(loader):
    probs = []
    with torch.no_grad():
        for xb, _ in loader:
            probs.extend(torch.sigmoid(final_model(xb.to(device))).cpu().numpy().tolist())
    return np.array(probs)

p_tr = get_preds(DataLoader(TensorDataset(torch.tensor(X_tr_fin), torch.tensor(y_tr_fin)), batch_size=256, shuffle=False))
p_va = get_preds(va_loader)
p_te = get_preds(te_loader)

# Tune optimal F1 threshold strictly on the validation set
prec_va, rec_va, thresh_va = precision_recall_curve(y_va_fin, p_va)
f1s_va = (2 * prec_va[:-1] * rec_va[:-1]) / (prec_va[:-1] + rec_va[:-1] + 1e-12)
best_th = float(thresh_va[np.argmax(f1s_va)]) if len(thresh_va) > 0 else 0.5
# 3. Test Evaluation & Plotting
# Calculate Train and Test Predictions using the validation-tuned threshold
y_tr_pred = (p_tr >= best_th).astype(int)
y_te_pred = (p_te >= best_th).astype(int)

print(f"\n✓ Final Model Performance Metrics (Threshold: {best_th:.2f}):")
print(f"  Training   | PR-AUC: {average_precision_score(y_tr_fin, p_tr):.4f} | ROC-AUC: {roc_auc_score(y_tr_fin, p_tr):.4f} | F1: {f1_score(y_tr_fin, y_tr_pred, zero_division=0):.4f}")
print(f"  Validation | PR-AUC: {average_precision_score(y_va_fin, p_va):.4f} | ROC-AUC: {roc_auc_score(y_va_fin, p_va):.4f} | F1: {np.max(f1s_va):.4f}")
print(f"  Unseen Test| PR-AUC: {average_precision_score(y_te_fin, p_te):.4f} | ROC-AUC: {roc_auc_score(y_te_fin, p_te):.4f} | F1: {f1_score(y_te_fin, y_te_pred, zero_division=0):.4f}")
fig, axes = plt.subplots(1, 3, figsize=(18, 5))

axes[0].plot(train_losses, label="Train Loss", lw=2, color="steelblue")
axes[0].plot(val_losses, label="Validation Loss", lw=2, color="darkorange")
axes[0].axvline(len(val_losses) - wait - 1, color='k', linestyle='--', label="Early Stopping Checkpoint")
axes[0].set_title("Training vs Validation Loss", fontweight="bold")
axes[0].set_xlabel("Epoch (0 = Random Weights)")
axes[0].set_ylabel("BCE Loss")
axes[0].legend()

te_prec, te_rec, _ = precision_recall_curve(y_te_fin, p_te)
axes[1].plot(te_rec, te_prec, color="purple", lw=2, label=f"Test PR-AUC = {average_precision_score(y_te_fin, p_te):.3f}")
axes[1].axhline(y_te_fin.mean(), color='r', linestyle='--', label="Baseline (Random)")
axes[1].set_title("Unseen Test Set: Precision-Recall Curve", fontweight="bold")
axes[1].set_xlabel("Recall")
axes[1].set_ylabel("Precision")
axes[1].legend()

cm = confusion_matrix(y_te_fin, y_te_pred)
disp = ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=["No Podium", "Podium"])
disp.plot(ax=axes[2], cmap="Blues", values_format="d")
axes[2].set_title(f"Test Confusion Matrix (Threshold = {best_th:.2f})", fontweight="bold")

plt.tight_layout()
plt.show()

# 4. Final Feature-Group Ablation (Validating Importance on Unseen Test Set)
print("\n" + "="*90)
print("FINAL MODEL FEATURE-GROUP ABLATION (Importance Validation on Unseen Test Set)")
print("Note: Explanations must not be presented as causal effects.")
print("="*90)

abl_results = []
base_te_pr = average_precision_score(y_te_fin, p_te)

for grp_name, cols in BEST_FEATURE_GROUPS.items():
    abl_cols = [c for g, clist in BEST_FEATURE_GROUPS.items() if g != grp_name for c in final_groups[g] if c in df_final.columns]
    
    X_tr_a = final_pipe.fit_transform(tr_final[abl_cols]).astype(np.float32)
    X_te_a = final_pipe.transform(te_final[abl_cols]).astype(np.float32)
    
    abl_metrics = train_eval_ffnn_quick(X_tr_a, y_tr_fin, X_te_a, y_te_fin)
    drop = base_te_pr - abl_metrics['Val_PR_AUC']  
    
    abl_results.append({
        "Ablated Group": grp_name,
        "Test PR-AUC Without Group": abl_metrics['Val_PR_AUC'],
        "Δ Test PR-AUC Drop": drop
    })

abl_df = pd.DataFrame(abl_results).sort_values(by="Δ Test PR-AUC Drop", ascending=False).reset_index(drop=True)
abl_df.index = abl_df.index + 1
abl_df.index.name = "Importance Rank"
display(abl_df.round(4))
```

# Cell 13: Explainable AI (XAI) - Global and Local Explanations

In this section, we apply Explainable AI (XAI) techniques to understand the decision-making process of our 4 trained models:
1. **Logistic Regression (Baseline)**
2. **Random Forest (Untuned)**
3. **Random Forest (Tuned)**
4. **Shallow Feedforward Neural Network (PyTorch)**

### 1. Global Explanations
* **Permutation Feature Importance:** Evaluates the decrease in model performance (PR-AUC) when a single feature is randomly shuffled. This breaks the relationship between the feature and the target, revealing how heavily the model relies on it.
* **SHAP (SHapley Additive exPlanations):** Provides a theoretically sound global summary of feature impacts. The summary plots show both the magnitude and direction of a feature's effect on the predicted probability.

### 2. Local Explanations
* **LIME (Local Interpretable Model-agnostic Explanations):** Explains a single, specific prediction by fitting a surrogate linear model locally around the instance. We use this to explain a specific driver's podium prediction.

*Important Note:* Explanations highlight correlational importance and model reliance, but they must not be interpreted as strict causal effects.

```python
# ==============================================================================
# CELL 13: EXPLAINABILITY (XAI) FOR ALL 4 MODELS
#   Global : Permutation Importance + SHAP (beeswarm & mean |SHAP|)
#   Local  : LIME + local SHAP waterfall for ONE test-set driver-race
# Models : Logistic Regression | Random Forest (untuned) | Random Forest (tuned) | Shallow FFNN
# NOTE   : Explanations describe what the MODEL relies on - they are NOT causal effects.
# ==============================================================================
import subprocess, sys, warnings
try:
    import shap
    import lime.lime_tabular as lime_tabular
except ImportError:
    subprocess.check_call([sys.executable, "-m", "pip", "install", "-q", "shap", "lime"])
    import shap
    import lime.lime_tabular as lime_tabular

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import torch
from scipy.stats import spearmanr
from sklearn.metrics import average_precision_score
from IPython.display import display

warnings.filterwarnings("ignore")
RNG_SEED = 42
rng = np.random.default_rng(RNG_SEED)
N_REPEATS = 5          # permutation repeats
N_SHAP_ROWS = 300      # test rows explained by SHAP (tree / linear models)
N_SHAP_ROWS_NN = 100   # KernelExplainer is slower -> smaller subset for the FFNN
TOP_K = 10             # features shown in each plot

# ------------------------------------------------------------------------------
# 1. Collect the 4 models with THEIR OWN matrices / feature names
#    (statistical models: X_train / X_test, `final_features`; FFNN: X_*_fin, `active_cols`)
# ------------------------------------------------------------------------------
def ffnn_predict(X):
    """Podium probability from the final Shallow FFNN (sigmoid of logits)."""
    final_model.eval()
    with torch.no_grad():
        logits = final_model(torch.tensor(np.asarray(X), dtype=torch.float32).to(device))
        return torch.sigmoid(logits).cpu().numpy()

MODELS = {}
for name, mdl in fitted_models.items():                      # LR, RF untuned, RF tuned (Cell 8)
    MODELS[name] = dict(
        kind="linear" if "Logistic" in name else "tree",
        predict=(lambda X, m=mdl: m.predict_proba(X)[:, 1]),
        model=mdl, Xtr=np.asarray(X_train), Xte=np.asarray(X_test),
        yte=np.asarray(y_test), feats=list(final_features), frame=test_df.reset_index(drop=True))
MODELS["Shallow FFNN (64-32)"] = dict(
    kind="nn", predict=ffnn_predict, model=final_model,
    Xtr=np.asarray(X_tr_fin), Xte=np.asarray(X_te_fin), yte=np.asarray(y_te_fin),
    feats=list(active_cols), frame=te_final.reset_index(drop=True))

print("Models explained:", list(MODELS))
for n, c in MODELS.items():
    print(f"  - {n:<42s} features={len(c['feats']):>2d} | test rows={len(c['yte'])}")

# ------------------------------------------------------------------------------
# 2. GLOBAL (a): Permutation Importance on the unseen test set (metric = PR-AUC)
#    importance_j = PR-AUC(original) - PR-AUC(feature j shuffled), averaged over repeats
# ------------------------------------------------------------------------------
def permutation_importance_prauc(predict, X, y, feats, n_repeats=N_REPEATS):
    base = average_precision_score(y, predict(X))
    means, stds = [], []
    for j in range(X.shape[1]):
        drops = []
        for _ in range(n_repeats):
            Xp = X.copy()
            Xp[:, j] = rng.permutation(Xp[:, j])
            drops.append(base - average_precision_score(y, predict(Xp)))
        means.append(np.mean(drops)); stds.append(np.std(drops))
    return pd.DataFrame({"feature": feats, "perm_importance": means, "std": stds}) \
             .sort_values("perm_importance", ascending=False).reset_index(drop=True), base

PERM = {}
for name, c in MODELS.items():
    PERM[name], base_pr = permutation_importance_prauc(c["predict"], c["Xte"], c["yte"], c["feats"])
    print(f"[Permutation] {name}: baseline Test PR-AUC = {base_pr:.4f}")

fig, axes = plt.subplots(2, 2, figsize=(16, 10))
for ax, (name, pdf) in zip(axes.ravel(), PERM.items()):
    top = pdf.head(TOP_K).iloc[::-1]
    ax.barh(top["feature"], top["perm_importance"], xerr=top["std"], color="#0f766e",
            edgecolor="black", capsize=3)
    ax.axvline(0, color="k", lw=0.8)
    ax.set_title(name, fontweight="bold")
    ax.set_xlabel("Drop in Test PR-AUC when feature is shuffled")
plt.suptitle("Global Explanation 1: Permutation Importance (Unseen Test Set 2022-2024)\n"
             "Model reliance, NOT causal effect", fontsize=14, fontweight="bold")
plt.tight_layout()
plt.show()

# ------------------------------------------------------------------------------
# 3. GLOBAL (b): SHAP  (Linear -> LinearExplainer | RF -> TreeExplainer | FFNN -> KernelExplainer)
#    SHAP values are in log-odds for LR, in probability units for RF / FFNN.
# ------------------------------------------------------------------------------
def build_explainer(c):
    if c["kind"] == "linear":
        bg = c["Xtr"][rng.choice(len(c["Xtr"]), 100, replace=False)]
        return shap.LinearExplainer(c["model"], bg)
    if c["kind"] == "tree":
        return shap.TreeExplainer(c["model"])
    bg = shap.kmeans(c["Xtr"], 30)                             # summarised background for Kernel SHAP
    return shap.KernelExplainer(c["predict"], bg)

def shap_for_class1(explainer, X, kind):
    if kind == "tree":
        sv = explainer.shap_values(X, check_additivity=False)
    elif kind == "nn":
        sv = explainer.shap_values(X, nsamples=300, silent=True)
    else:
        sv = explainer.shap_values(X)
    if isinstance(sv, list):             # old shap: [class0, class1]
        sv = sv[1]
    sv = np.asarray(sv)
    if sv.ndim == 3:                     # new shap: (n, F, n_classes) or (n, F, 1)
        sv = sv[:, :, -1]
    return sv

def base_value(explainer):
    ev = np.atleast_1d(explainer.expected_value)
    return float(ev[-1])

SHAP = {}
shap_idx = rng.choice(len(MODELS["Shallow FFNN (64-32)"]["yte"]), N_SHAP_ROWS, replace=False)
for name, c in MODELS.items():
    n_rows = N_SHAP_ROWS_NN if c["kind"] == "nn" else N_SHAP_ROWS
    idx = shap_idx[:n_rows] if len(c["Xte"]) >= shap_idx.max() + 1 else \
          rng.choice(len(c["Xte"]), n_rows, replace=False)
    Xs = c["Xte"][idx]
    expl = build_explainer(c)
    sv = shap_for_class1(expl, Xs, c["kind"])
    SHAP[name] = dict(explainer=expl, sv=sv, X=Xs, idx=idx)
    print(f"[SHAP] {name}: explained {len(Xs)} test rows -> values shape {sv.shape}")

# 3a. Beeswarm summary plot per model (colour = standardised feature value: red high, blue low)
for name, c in MODELS.items():
    s = SHAP[name]
    plt.figure(figsize=(9, 6))
    shap.summary_plot(s["sv"], s["X"], feature_names=c["feats"], max_display=TOP_K, show=False)
    plt.title(f"Global Explanation 2: SHAP Beeswarm - {name}", fontweight="bold")
    plt.tight_layout()
    plt.show()

# 3b. Mean |SHAP| bar plot, all 4 models side by side
fig, axes = plt.subplots(2, 2, figsize=(16, 10))
SHAP_IMP = {}
for ax, (name, c) in zip(axes.ravel(), MODELS.items()):
    imp = pd.Series(np.abs(SHAP[name]["sv"]).mean(axis=0), index=c["feats"]).sort_values(ascending=False)
    SHAP_IMP[name] = imp
    imp.head(TOP_K).iloc[::-1].plot(kind="barh", ax=ax, color="#7c3aed", edgecolor="black")
    ax.set_title(name, fontweight="bold")
    ax.set_xlabel("Mean |SHAP value|")
plt.suptitle("Global Explanation 2: Mean |SHAP| Feature Importance", fontsize=14, fontweight="bold")
plt.tight_layout()
plt.show()

# 3c. Compare the two global methods: top-5 features + rank agreement per model
rows = []
for name in MODELS:
    perm_s = PERM[name].set_index("feature")["perm_importance"]
    shap_s = SHAP_IMP[name]
    common = perm_s.index.intersection(shap_s.index)
    rho = spearmanr(perm_s[common], shap_s[common])[0]
    rows.append({"Model": name,
                 "Top-5 Permutation": ", ".join(PERM[name]["feature"].head(5)),
                 "Top-5 mean |SHAP|": ", ".join(shap_s.index[:5]),
                 "Spearman(Perm, SHAP)": round(rho, 3)})
print("\n=== Global explanation summary: do the two methods agree? ===")
display(pd.DataFrame(rows))

# ------------------------------------------------------------------------------
# 4. LOCAL explanations (SHAP waterfall + LIME) for ONE unseen-test driver-race
#    Case: a TRUE podium finisher whose tuned-RF probability is closest to 0.5
#    (an "uncertain" prediction is more informative to explain than an obvious pole-sitter).
# ------------------------------------------------------------------------------
ref = MODELS["Random Forest (Tuned: max_depth=6)"]
p_ref = ref["predict"](ref["Xte"])
cand = np.where(ref["yte"] == 1)[0]
i_ref = cand[np.argmin(np.abs(p_ref[cand] - 0.5))]
case = ref["frame"].iloc[i_ref]
race_id, driver_id = case["raceId"], case["driverId"]

drv = f1_data["drivers"].set_index("driverId")
rac = f1_data["races"].set_index("raceId")
drv_name = f"{drv.loc[driver_id, 'forename']} {drv.loc[driver_id, 'surname']}"
race_name = f"{int(rac.loc[race_id, 'year'])} {rac.loc[race_id, 'name']}"
print(f"\nLocal case: {drv_name} - {race_name} | grid = {case['grid']} | actual podium = 1")

# locate the same driver-race row in each model's own test matrix
local_pos = {}
for name, c in MODELS.items():
    fr = c["frame"]
    pos = np.where((fr["raceId"] == race_id) & (fr["driverId"] == driver_id))[0]
    local_pos[name] = int(pos[0])

summary_rows = []
for name, c in MODELS.items():
    x = c["Xte"][local_pos[name]]
    summary_rows.append({"Model": name, "Predicted podium probability": round(float(c["predict"](x.reshape(1, -1))[0]), 4),
                         "Actual": "Podium"})
display(pd.DataFrame(summary_rows))

# 4a. Local SHAP waterfall per model
for name, c in MODELS.items():
    x = c["Xte"][local_pos[name]].reshape(1, -1)
    expl = SHAP[name]["explainer"]
    sv1 = shap_for_class1(expl, x, c["kind"])[0]
    exp_obj = shap.Explanation(values=sv1, base_values=base_value(expl), data=x[0], feature_names=c["feats"])
    shap.plots.waterfall(exp_obj, max_display=TOP_K, show=False)
    plt.gcf().suptitle(f"Local SHAP - {name}\n{drv_name}, {race_name}", fontweight="bold", y=1.02)
    plt.show()

# 4b. Local LIME per model (2x2 panel)
fig, axes = plt.subplots(2, 2, figsize=(16, 10))
for ax, (name, c) in zip(axes.ravel(), MODELS.items()):
    lime_exp = lime_tabular.LimeTabularExplainer(
        training_data=c["Xtr"], feature_names=c["feats"], class_names=["No Podium", "Podium"],
        mode="classification", discretize_continuous=True, random_state=RNG_SEED)
    predict_proba2 = lambda X, f=c["predict"]: np.column_stack([1 - f(X), f(X)])
    exp = lime_exp.explain_instance(c["Xte"][local_pos[name]], predict_proba2,
                                    num_features=8, num_samples=2000)
    items = exp.as_list()[::-1]
    ax.barh([k for k, _ in items], [v for _, v in items],
            color=["#16a34a" if v > 0 else "#dc2626" for _, v in items], edgecolor="black")
    ax.axvline(0, color="k", lw=0.8)
    ax.set_title(f"{name}\nLIME local fit R2 = {exp.score:.2f}", fontweight="bold")
    ax.set_xlabel("LIME weight (green pushes toward Podium, red toward No Podium)")
plt.suptitle(f"Local Explanation: LIME - {drv_name}, {race_name}", fontsize=14, fontweight="bold")
plt.tight_layout()
plt.show()

print("\n" + "=" * 90)
print("IMPORTANT: Permutation importance, SHAP and LIME describe which features each MODEL relies on")
print("to make its predictions. They are NOT causal effects: e.g. a high importance for `grid` does")
print("not prove that starting position causes podiums (it also proxies car quality, qualifying pace,")
print("constructor strength). Correlated features (grid / position / q1-q3 / standings) also share")
print("importance, so each one can look weaker under permutation than the group really is.")
print("=" * 90)
```

```python
import shap
import lime.lime_tabular
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import torch

# ==============================================================================
# Local Explainability (XAI) for ALL 4 MODELS
# ==============================================================================

# 1. Define a helper for the FFNN prediction
def ffnn_predict(X):
    final_model.eval()
    with torch.no_grad():
        logits = final_model(torch.tensor(np.asarray(X), dtype=torch.float32).to(device))
        return torch.sigmoid(logits).cpu().numpy()

# 2. Package all 4 models, their specific test data, and their EXACT feature names
MODELS = {}

# A. The 3 Statistical Models (using prep_pipeline transformed data)
for name, mdl in fitted_models.items(): 
    MODELS[name] = {
        "model": mdl,
        "predict_fn": lambda X, m=mdl: m.predict_proba(X)[:, 1],
        "X_train": np.asarray(X_train),
        "X_test": np.asarray(X_test),
        "feature_names": final_features, # Actual feature names!
        "is_nn": False
    }

# B. The PyTorch FFNN (using final_pipe transformed data)
MODELS["Shallow FFNN (64-32)"] = {
    "model": final_model,
    "predict_fn": ffnn_predict,
    "X_train": np.asarray(X_tr_fin),
    "X_test": np.asarray(X_te_fin),
    "feature_names": active_cols, # Actual feature names!
    "is_nn": True
}

# 3. Select a specific instance to explain (e.g., index 0 from the test set)
instance_idx = 0

for name, model_info in MODELS.items():
    print(f"\n" + "="*80)
    print(f"Generating Local Explanations for: {name}")
    print("="*80)
    
    X_tr = model_info["X_train"]
    X_te = model_info["X_test"]
    feats = model_info["feature_names"]
    single_instance = X_te[instance_idx]
    
    # ---------------------------------------------------------
    # SHAP EXPLANATION
    # ---------------------------------------------------------
    if model_info["is_nn"]:
        # FFNN requires KernelExplainer
        bg = shap.kmeans(X_tr, 30) 
        explainer = shap.KernelExplainer(model_info["predict_fn"], bg)
        shap_values = explainer.shap_values(single_instance.reshape(1, -1), nsamples=300, silent=True)
        # KernelExplainer returns an array, build Explanation object manually
        base_val = float(np.atleast_1d(explainer.expected_value)[-1])
        val = shap_values[0] if isinstance(shap_values, list) else shap_values[0]
        exp_obj = shap.Explanation(values=val, base_values=base_val, data=single_instance, feature_names=feats)
    elif "Random Forest" in name:
        # RF uses TreeExplainer
        explainer = shap.TreeExplainer(model_info["model"])
        shap_values = explainer.shap_values(single_instance.reshape(1, -1), check_additivity=False)
        base_val = float(np.atleast_1d(explainer.expected_value)[-1])
        # Extract Class 1
        val = shap_values[1][0] if isinstance(shap_values, list) else shap_values[0, :, 1]
        exp_obj = shap.Explanation(values=val, base_values=base_val, data=single_instance, feature_names=feats)
    else:
        # Logistic Regression uses LinearExplainer
        bg = X_tr[np.random.choice(X_tr.shape[0], 100, replace=False)]
        explainer = shap.LinearExplainer(model_info["model"], bg)
        shap_values = explainer.shap_values(single_instance.reshape(1, -1))
        base_val = float(np.atleast_1d(explainer.expected_value)[-1])
        exp_obj = shap.Explanation(values=shap_values[0], base_values=base_val, data=single_instance, feature_names=feats)

    plt.figure(figsize=(8, 4))
    plt.title(f"Local SHAP Waterfall - {name}", fontweight="bold", y=1.05)
    shap.plots.waterfall(exp_obj, max_display=10, show=False)
    plt.show()

    # ---------------------------------------------------------
    # LIME EXPLANATION
    # ---------------------------------------------------------
    # LIME expects a function that outputs probabilities for BOTH classes [Prob(0), Prob(1)]
    def lime_predict_wrapper(X, p_fn=model_info["predict_fn"]):
        p1 = p_fn(X)
        return np.column_stack([1 - p1, p1])

    lime_explainer = lime.lime_tabular.LimeTabularExplainer(
        training_data=X_tr,
        feature_names=feats,
        class_names=["No Podium", "Podium"],
        mode="classification",
        random_state=42
    )
    
    lime_exp = lime_explainer.explain_instance(
        data_row=single_instance,
        predict_fn=lime_predict_wrapper,
        num_features=5
    )
    
    lime_exp.show_in_notebook(show_table=True)
```

# Model Behavior & Known Limitations
---
While our predictive pipelines successfully identify key podium signals, they are constrained by several architectural and domain-specific limitations:

* **Logistic Regression:** Assumes a strictly linear relationship between features (like grid position) and the log-odds of a podium. It cannot natively capture non-linear interactions (e.g., grid position being more decisive at Monaco than at Monza) without manual feature crossing.
* **Random Forest (Tuned):** While it effectively captures non-linear interactions (like qualifying pace vs. constructor reliability), it cannot extrapolate trends outside historical bounding boxes. Because it relies on historical splits, it is sensitive to sudden regulation resets (like the 2022 Ground-Effect era), which can degrade performance until it learns the new pecking order.
* **Shallow FFNN:** Suffers from "Historical Qualifying Coverage Drift." Because pre-2006 races lack Q1/Q2/Q3 knockout times, these missing values are median-imputed, which compresses the neural network's ability to learn from single-lap pace differences in older eras. 
* **Global Limitation (All Models):** Because the models strictly obey a pre-race cutoff to avoid target leakage, they are completely blind to stochastic Sunday race-day events (Lap 1 collisions, sudden rain, mid-race safety cars, or sudden mechanical failures), establishing a hard ceiling on predictive accuracy.
